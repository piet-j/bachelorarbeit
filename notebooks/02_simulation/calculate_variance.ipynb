```python
import pandas as pd

def estimate_v_per_season(long_df):
    # Finale Tordifferenz pro Team und Saison
    final_gd = (
        long_df
        .groupby(["season_id", "team_id"])["X"]
        .sum()
    )

    # Geschätzte Teamstärken pro Spiel (38 Spiele pro Saison)
    estimated_teamstrengths = final_gd / 38  

    # Varianz der Leistungsstärken pro Saison (Varianz der finalen Tordifferenzen)
    # Mittelwert der Leistungsstärken ist 0
    var_per_season = (estimated_teamstrengths**2).groupby("season_id").mean()

    # Ergebnis als DataFrame zur besseren Lesbarkeit
    result = pd.DataFrame({
        "var_per_season": var_per_season,
    })
    return result


# ---- Ausführung ----
long_df = pd.read_csv("") # Pfad des Profifußball Long-Formats angeben
v_table = estimate_v_per_season(long_df)

# Ausgabe pro Saison
print("v pro Saison:")
print(v_table)

# Mittelwert über alle Saisons
v_mean = v_table["var_per_season"].mean()
print(f"\nMittelwert von v über alle Saisons: {v_mean:.4f}")
```
